```python
df_dedup = df.groupby(['Commodity', 'Arrival_Date'], as_index=False).agg({
    'Modal_Price': 'mean',
    'Min_Price': 'mean',
    'Max_Price': 'mean',
    'rainfall_mm': 'first',
    'rainfall_7day_sum': 'first',
    'temp_max': 'first',
    'temp_min': 'first',
    'days_to_nearest_festival': 'first',
    'is_festival_week': 'first'
})

print(df.shape, "->", df_dedup.shape)
```

```python
def add_calendar_aware_lags(group):
    group = group.set_index('Arrival_Date').sort_index()
    full_range = pd.date_range(group.index.min(), group.index.max(), freq='D')
    group = group.reindex(full_range)
    
    for lag in [1, 7, 14, 30]:
        group[f'price_lag_{lag}'] = group['Modal_Price'].shift(lag)
    
    return group
```

```python
lagged = df_dedup.groupby('Commodity', group_keys=True).apply(add_calendar_aware_lags, include_groups=False)
lagged = lagged.reset_index().rename(columns={'level_1': 'Arrival_Date'})
lagged = lagged.dropna(subset=['Modal_Price'])

print(lagged.shape)
lagged[['Commodity', 'Arrival_Date', 'Modal_Price', 'price_lag_1', 'price_lag_7', 'price_lag_14', 'price_lag_30']].head(10)
```

```python
model_df = pd.read_csv("data/processed/surat_model_ready.csv")
model_df['Arrival_Date'] = pd.to_datetime(model_df['Arrival_Date'])

print(model_df.shape)
print(model_df.columns.tolist())
print(model_df.isna().sum())
```

```python
from sklearn.preprocessing import OneHotEncoder

model_df = model_df.sort_values('Arrival_Date').reset_index(drop=True)

encoder = OneHotEncoder(sparse_output=False, handle_unknown='ignore')
commodity_encoded = encoder.fit_transform(model_df[['Commodity']])

# Turn the encoded array into a DataFrame with proper column names
encoded_cols = encoder.get_feature_names_out(['Commodity'])
commodity_df = pd.DataFrame(commodity_encoded, columns=encoded_cols, index=model_df.index)

# Combine back with the rest of the data
model_df = pd.concat([model_df.drop(columns=['Commodity']), commodity_df], axis=1)

print(model_df.shape)
model_df.columns.tolist()
```

```python
train_df = model_df[model_df['Arrival_Date'] < '2024-01-01']
test_df = model_df[model_df['Arrival_Date'] >= '2024-01-01']

print("Train:", train_df.shape, "| Date range:", train_df['Arrival_Date'].min(), "to", train_df['Arrival_Date'].max())
print("Test:", test_df.shape, "| Date range:", test_df['Arrival_Date'].min(), "to", test_df['Arrival_Date'].max())
```

```python
drop_cols = ['Arrival_Date', 'Modal_Price', 'Min_Price', 'Max_Price']  # Modal_Price is our target, Arrival_Date isn't a usable numeric feature directly

X_train = train_df.drop(columns=drop_cols)
y_train = train_df['Modal_Price']

X_test = test_df.drop(columns=drop_cols)
y_test = test_df['Modal_Price']

print(X_train.shape, X_test.shape)
```

```python
from xgboost import XGBRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, mean_absolute_percentage_error , r2_score
import numpy as np

model = XGBRegressor(
    n_estimators=200,
    max_depth=6,
    learning_rate=0.05,
    random_state=42
)

model.fit(X_train, y_train)

y_pred = model.predict(X_test)

mae = mean_absolute_error(y_test, y_pred)
rmse = np.sqrt(mean_squared_error(y_test, y_pred))
mape = mean_absolute_percentage_error(y_test, y_pred)
r2 =  r2_score(y_test, y_pred)

print(f"MAE: {mae:.2f}")
print(f"RMSE: {rmse:.2f}")
print(f"MAPE: {mape:.2%}")
print(f"r2 :{r2}")
```

```python
valid_mask = X_test['price_lag_1'].notna()

comparison_valid = pd.DataFrame({
    'Arrival_Date': test_df.loc[valid_mask, 'Arrival_Date'].values,
    'actual': y_test[valid_mask].values,
    'predicted': y_pred[valid_mask],
    'baseline': X_test.loc[valid_mask, 'price_lag_1'].values
})

commodity_cols = [c for c in test_df.columns if c.startswith('Commodity_')]
comparison_valid['Commodity'] = test_df.loc[valid_mask, commodity_cols].idxmax(axis=1).str.replace('Commodity_', '').values

comparison_valid['model_abs_error'] = abs(comparison_valid['actual'] - comparison_valid['predicted'])
comparison_valid['baseline_abs_error'] = abs(comparison_valid['actual'] - comparison_valid['baseline'])

comparison_valid['model_pct_error'] = comparison_valid['model_abs_error'] / comparison_valid['actual']
comparison_valid['baseline_pct_error'] = comparison_valid['baseline_abs_error'] / comparison_valid['actual']

per_commodity_comparison = comparison_valid.groupby('Commodity').agg(
    model_mae=('model_abs_error', 'mean'),
    baseline_mae=('baseline_abs_error', 'mean'),
    model_mape=('model_pct_error', 'mean'),
    baseline_mape=('baseline_pct_error', 'mean')
).round(3)

per_commodity_comparison['model_wins'] = per_commodity_comparison['model_mape'] < per_commodity_comparison['baseline_mape']

print(per_commodity_comparison.sort_values('model_mape'))
```

```python
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor
from xgboost import XGBRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, mean_absolute_percentage_error
import numpy as np

models = {
    "Decision Tree": DecisionTreeRegressor(max_depth=8, random_state=42),
    "Random Forest": RandomForestRegressor(n_estimators=200, max_depth=8, random_state=42, n_jobs=-1),
    "XGBoost": XGBRegressor(n_estimators=200, max_depth=6, learning_rate=0.05, random_state=42)
}

results = []

for name, m in models.items():
    m.fit(X_train, y_train)
    preds = m.predict(X_test)
    
    mae = mean_absolute_error(y_test, preds)
    rmse = np.sqrt(mean_squared_error(y_test, preds))
    mape = mean_absolute_percentage_error(y_test, preds)
    
    results.append({"Model": name, "MAE": mae, "RMSE": rmse, "MAPE": mape})
    print(f"{name}: MAE={mae:.2f}, RMSE={rmse:.2f}, MAPE={mape:.2%}")

results_df = pd.DataFrame(results)
print("\n", results_df)
```
